## Network Capacity

Add infrastructure capacity to graph.

```python
# Set parent as root and import config
import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parent))
import configs.simulation_config as cfg
from configs.paths import PROCESSED_DIR

import osmnx as ox
import pandas as pd
```

```python
def add_capacity(edges_df, lane_width, default_lanes):
    edges_df = edges_df.copy()

    lanes_numeric = pd.to_numeric(edges_df.get("lanes"), errors="coerce")
    lanes_numeric = lanes_numeric.fillna(default_lanes).clip(lower=default_lanes)

    edges_df["lanes_clean"] = lanes_numeric
    edges_df["capacity"] = edges_df["length"].fillna(0) * lane_width * edges_df["lanes_clean"]

    return edges_df
```

```python
graph = ox.load_graphml(PROCESSED_DIR / "hatyai_graph_clean.graphml")
nodes, edges = ox.graph_to_gdfs(graph)

edges = add_capacity(edges, cfg.LANE_WIDTH_M, cfg.DEFAULT_LANES)
graph = ox.graph_from_gdfs(nodes, edges)
ox.save_graphml(graph, PROCESSED_DIR / "hatyai_graph_with_capacity.graphml")
```
