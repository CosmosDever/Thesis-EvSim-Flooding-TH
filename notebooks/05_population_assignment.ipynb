## Population assignment

Attach demographic load to network.

```python
# Set parent as root and import config
import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parent))
import configs.simulation_config as cfg

import numpy as np
import osmnx as ox
```

```python
def assign_population(nodes_df, pop_male, pop_female, households, seed=888):
    rng = np.random.default_rng(seed)
    n = len(nodes_df)
    weights = np.ones(n) / n

    nodes_df["pop_male"] = rng.multinomial(pop_male, weights)
    nodes_df["pop_female"] = rng.multinomial(pop_female, weights)
    nodes_df["pop_total"] = nodes_df["pop_male"] + nodes_df["pop_female"]
    nodes_df["households"] = rng.multinomial(households, weights)

    return nodes_df
```

```python
graph = ox.load_graphml("../data/processed/hatyai_graph_with_shelters.graphml")
nodes, edges = ox.graph_to_gdfs(graph)

assign_population(
    nodes, 
    cfg.MALE_POPULATION, 
    cfg.FEMALE_POPULATION, 
    cfg.TOTAL_POPULATION, 
    cfg.HOUSEHOLDS)

graph = ox.graph_from_gdfs(nodes, edges)
ox.save_graphml(graph, "../data/processed/hatyai_graph_ready_for_flood.graphml")
```
